```python
import numpy as np
import pandas as pd
```

```python
weight_summary = pd.read_csv('data/2025-06-19-10_weights_summary.csv')
weight_summary
```

```python
# check which weight norms are greater than 1
weight_summary['norm_gt_1'] = weight_summary['norm'] > 1
# filter the DataFrame to show only those rows
weight_summary_gt_1 = weight_summary[weight_summary['norm_gt_1']]
weight_summary_gt_1.sort_values('norm', ascending=False)
```

```python
# get rows where name contains attn_
weight_summary_attn = weight_summary[weight_summary['name'].str.contains('attn_')]
# sort by norm in descending order
weight_summary_attn.sort_values('norm', ascending=False)
```

```python
# get rows where name contains pTo
weight_summary_attn = weight_summary[weight_summary['name'].str.contains('pTo')]
# sort by norm in descending order
weight_summary_attn.sort_values('norm', ascending=False)
```
