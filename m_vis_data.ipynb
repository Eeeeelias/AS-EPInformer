```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
psi_response = pd.read_csv('data/psi_response.csv')
psi_response
```

```python
# plot distribution of K562_SE_psi
plt.figure(figsize=(10, 6))
sns.histplot(psi_response[psi_response['event_id'].str.contains(';SE:')]['K562_SE_psi'], bins=30, kde=True)
plt.title('Distribution of K562_SE_psi')
plt.xlabel('K562_SE_psi')
plt.ylabel('Frequency')
plt.grid()
plt.show()
```

## visualise the training sizes for each fold

```python
split_sizes = pd.read_csv('data/data_split_sizes.csv')
split_sizes
```

```python
# melt the table to have a long format for seaborn
split_sizes_melted = split_sizes.melt(id_vars=['fold'], var_name='dataset', value_name='size')
split_sizes_melted
```

```python
# plot data split sizes
plt.figure(figsize=(10, 6))
sns.barplot(data=split_sizes_melted, x='fold', y='size', hue='dataset')
plt.title('Data Split Sizes by Fold and Dataset')
plt.xlabel('Fold')
plt.ylabel('Size')
```
